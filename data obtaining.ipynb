```python
import urllib.request
import json                 # Used to load data into JSON format
from pprint import pprint   # pretty-print
```

```python
import pandas as pd

frames = []
for start in [1, 10001, 20001, 30001]:
    count = 10000 if start < 30001 else 6454
    url = f"https://aoe2.net/api/leaderboard?game=aoe2de&leaderboard_id=3&start={start}&count={count}"
    response = urllib.request.urlopen(url)
    text = response.read()
    json_data = json.loads(text)
    
    df = pd.DataFrame(data=json_data['leaderboard'])
    df = df.set_index('rank')
    frames.append(df)

leaderboard = pd.concat(frames)
```
